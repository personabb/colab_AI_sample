## chapter10 LangChainの自由度を体感する【RunnableLambda】

8セル目まで実行してください。

```python
#2セル目

from langchain_core.runnables import RunnableLambda

# ここでは、入力の文章に「技術」か「料理」かが入っているかどうかを判定するような関数を定義する
def check_tech_or_cooking(input_text: str) -> str:
    if "技術" in input_text:
        return "技術"
    elif "料理" in input_text:
        return "料理"
    else:
        return "その他"


runnnable = RunnableLambda(check_tech_or_cooking)
print("====RunnableLambdaの出力====")
print(runnnable.invoke("私は料理が得意です。"))
print(runnnable.invoke("私は技術が得意です。"))
print(runnnable.invoke("私は何も得意ではありません。"))
```

```python
#3セル目

from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.runnables import RunnableLambda

def make_prompt(input_text: str) -> list:

    message = [
        SystemMessage(content="あなたは日本語を話す優秀なアシスタントです。回答には必ず日本語で答えてください。また考える過程も出力してください。"),
        HumanMessage(
            content=[
                {
                    "type": "text",
                    "text": input_text,
                }
            ]
        )
    ]
    return message

runnnable = RunnableLambda(make_prompt)
print("====RunnableLambdaの出力====")
print(runnnable.invoke("仕事の熱意を取り戻すためのアイデアを5つ挙げてください。"))

```

```python
#4セル目

from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI

# geminiモデルを定義する
model = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    temperature=0,
)


# チェーンを定義する
#chain = RunnableLambda(make_prompt) | model | StrOutputParser()
chain = make_prompt | model | StrOutputParser()

print("====modelの出力====")
print(chain.invoke("仕事の熱意を取り戻すためのアイデアを5つ挙げてください。"))
```

```python
#5セル目

from langchain_core.runnables import RunnableLambda

def upper(input_text: str) -> str:
    return input_text.upper()

runnnable = RunnableLambda(upper)
print("====RunnableLambdaの出力====")
print(runnnable.invoke("Hello, world!"))
```

```python
#6セル目

from langchain_core.messages import HumanMessage

message = [HumanMessage(content="Hi there! How are you?")]

# チェーンを定義する
chain =  model | StrOutputParser() | RunnableLambda(upper)
print("====modelの出力====")
print(chain.invoke(message))
```

```python
#7セル目

"""upper()の出力は文字列型である必要があるため、この形は本来エラーになる"""
chain2 =  model  | RunnableLambda(upper)

print("====modelの出力====")

"""入出力の型が異なる場合、エラーになる"""
print(chain2.invoke(message))
```

```python
#8セル目

from langchain_core.messages import BaseMessage

#入出力を合わせた関数を新たに定義する
def model2upper(model_output:BaseMessage) -> str:
    return model_output.content.upper()

message = [HumanMessage(content="Hi there! How are you?")]
chain2 =  model | RunnableLambda(model2upper)

print("====modelの出力====")
print(chain2.invoke(message))
```

## chapter11 LLMの出力の形式を制限する

11セル目まで実行してください。

```python
#9セル目

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate,  HumanMessagePromptTemplate
from langchain_core.messages import  SystemMessage

system_prompt="""
回答する際は下記のように構造化して出力してください
- reasoning: ユーザの質問をステップバイステップで検討し、最終的な回答を推論し、全ての思考をこのフィールドに出力してください。
- conclusion: 最終的な回答のみを出力してください。
"""

# promptテンプレートを定義する
message = [
        SystemMessage(content= system_prompt),
        HumanMessagePromptTemplate.from_template(
            [
                {
                    "type": "text",
                    "text": "{question}"
                },
            ]
        )
    ]


prompt = ChatPromptTemplate.from_messages(message)


# geminiモデルを定義する
model = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    temperature=0,
)
```

```python
#10セル目

from pydantic import BaseModel, Field

# 出力の構造をモデルクラスとして定義する
class OutputModel(BaseModel):
    reasoning: str = Field(..., description='問題を解く上で必要な全ての思考内容と、最終的な結果を出力する')
    conclusion: str = Field(..., description='これまでの思考結果から最終的な結論のみを出力する')

chain = prompt | model.with_structured_output(OutputModel)


question = """
公園と学校の距離は1200mです。
A君が公園から、B君は学校から向かい合って同時に出発すると8分で出会いました。
また別の日、B君が学校から出た5分後にA君が追いかけるとA君が出発して10分後に追いつきました。
A君の速さは、「分速何m」でしょうか？
"""

result = chain.invoke({"question": question})

print("====modelの出力====")
print("===思考過程===")
print(result.reasoning)
print("===結論===")
print(result.conclusion)
```

```python
# 11セル目(答えだけを出力させた場合にどうなるか)
message = [
        SystemMessage(content= "答えだけを簡潔に出力してください。答え以外は出力を禁止します。"),
        HumanMessagePromptTemplate.from_template(
            [
                {
                    "type": "text",
                    "text": "{question}"
                },
            ]
        )
    ]

prompt = ChatPromptTemplate.from_messages(message)


# geminiモデルを定義する
model = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash",
    temperature=0,
)

chain = prompt | model

question = """
公園と学校の距離は1200mです。
A君が公園から、B君は学校から向かい合って同時に出発すると8分で出会いました。
また別の日、B君が学校から出た5分後にA君が追いかけるとA君が出発して10分後に追いつきました。
A君の速さは、「分速何m」でしょうか？
"""

print("====modelの出力====")
print(chain.invoke({"question": question}).content)

```
