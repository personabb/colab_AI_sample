## chapter4 プロンプトを構築する方法
3セル目まで実行してください。

```python
#2セル目

import os
from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate
from langchain_core.messages import SystemMessage
```

```python
#3セル目


# prompt構築の書き方1
message1 = [
        SystemMessage(content= "あなたは日本語を話す優秀なアシスタントです。回答には必ず日本語で答えてください。また考える過程も出力してください。"),
        HumanMessagePromptTemplate.from_template(
            [
                {
                    "type": "text",
                    "text": "{user_input}"
                },
            ]
        )
    ]

prompt1 = ChatPromptTemplate.from_messages(message1)

# prompt構築の書き方2
message2 = [
        ("system", "あなたは日本語を話す優秀なアシスタントです。回答には必ず日本語で答えてください。また考える過程も出力してください。"),
        ("human", "{user_input}"),
    ]

prompt2 = ChatPromptTemplate.from_messages(message2)
```

## chapter6 プロンプトとモデルを実行する方法
8セル目まで実行してください。

```python
#7セル目

llm_prompt1 = prompt1.invoke({"user_input": "こんにちは"})
llm_prompt2 = prompt2.invoke({"user_input": "こんにちは"})

print(llm_prompt1)
print(llm_prompt2)
```

```python
#8セル目

response1 = model.invoke(llm_prompt1)
response2 = model.invoke(llm_prompt2)

# 返り値の確認
print("====model output====")
print(response1)
print(response2)
print("====model output only text====")
# LLMが生成したテキストのみを確認
print(response1.content)
print(response2.content)

print("====model output with stream====")
# stream出力を試してみる。
llm_prompt3 = prompt1.invoke({"user_input": "日本で一番高い山について詳細に教えてください。"})
for chunk in model.stream(llm_prompt3):
    print(chunk.content, end="", flush=True)
```
